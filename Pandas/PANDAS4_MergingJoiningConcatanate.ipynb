#Merging 2 dataframes

```python
import numpy as np

import pandas as pd
```

```python
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DataFrame 2: Salary information
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees,salaries) #this will give common rows and merge them
```

### In merging **how = "inner"** means that it will give only common rows

### In merging **how = "outer"** means that it will give all the rows with NaN

```python
pd.merge(employees,salaries,on="employee_id",how="inner")
```

```python
df1=pd.merge(employees,salaries,on="employee_id",how="outer")
df1
```

```python
df1.fillna(value="Unknown")
```

##Concatenation of 2 dataframes

```python
daf1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
})

daf2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
})
```

```python
daf1
```

```python
pd.concat([daf1,daf2])  #this concates in columns wise
```

```python
pd.concat([daf1,daf2],axis =1) #this concatnates rows wise
```

#Joining 2 dataframes


```python
f1 = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie']
}, index=[1, 2, 3])

# Second DataFrame
f2 = pd.DataFrame({
    'score': [85, 90, 75]
}, index=[2, 3, 4])
```

```python
new_f1=f1.join(f2,how="outer")
new_f1.fillna(value=0)
```

```python
f2.join(f1,how="outer")
```
