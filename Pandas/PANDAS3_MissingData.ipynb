#Missing Data

```python
import numpy as np
import pandas as pd
```

##Finding missing data

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, 4, 5],
    'C': [1, np.nan, np.nan, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}                   #np.nan means NOT A NUMBER

dataf=pd.DataFrame(data)
dataf
```

```python
dataf.isna()
```

```python
dataf.isna().sum()
```

```python
dataf.isna().any()
```

##Removing the missing data

```python
dataf.dropna() #all the rows got deleted since all the rows had null values
```

###threshold

```python
dataf.dropna(thresh=3) # thresh counts non-missing values not missing values
```

##filling the missing values

```python
dataf.fillna(0) ## to fill the nan values we use (fillna)
```

```python
values ={"A":0,"B":2,"C":6,"D":8}  ##column wise filling
dataf.fillna(value=values)
```
