#Creating Dataframe

```python
import numpy as np
import pandas as pd
```

```python
employees = {
    "Name": ["Aarav", "Priya", "Rohan", "Sneha", "Arjun", "Meera"],
    "age": [24, 26, 23, 25, 27, 22],
    "city": ["Bangalore", "Mumbai", "Pune", "Hyderabad", "Chennai", "Delhi"],
    "salary": [55000, 62000, 48000, 58000, 70000, 45000]
}
```

```python
df=pd.DataFrame(employees) ## multiple series becomes dataframe
```

```python
df
```

```python
employees = [     ##if there are no keys given then it will default have index in table
    ["Aarav", 24, "Bangalore", 55000],
    ["Priya", 26, "Mumbai", 62000],
    ["Rohan", 23, "Pune", 48000],
    ["Sneha", 25, "Hyderabad", 58000],
    ["Arjun", 27, "Chennai", 70000],
    ["Meera", 22, "Delhi", 45000]
]

df2= pd.DataFrame(employees)
df2
```

```python
colm = ["Name","Age","City","Salary"]
df2 = pd.DataFrame(employees,columns=colm) ## adding columns manually
df2
```

##removing any column and row

##### only giving the name and axis wont delete the row or column from the main memeory it will just give a copy without that row or column,To delete it from main memory we have to use **[inplace=True**] then it will remove from the actual memory also.





```python
df2.drop("Name",axis=1,inplace=True) #delete any column by their name
```

```python
df2.drop(2,axis=0,inplace=True) #delete any row by indices
```

##Selecting Subsets of Rows and columns

```python
# we will make new dataframe3

Students = {
    "Name": ["Aarav", "Priya", "Rohan", "Sneha", "Arjun", "Meera"],
    "age": [24, 26, 23, 25, 27, 22],
    "city": ["Bangalore", "Mumbai", "Pune", "Hyderabad", "Chennai", "Delhi"],
    "salary": [55000, 62000, 48000, 58000, 70000, 45000]
}
```

```python
df3=pd.DataFrame(Students)
```

```python
df3
```

```python
df3.loc[[0,1],["city","salary"]] # using loc and then give the rows you want and columns you want to access
```

```python
df3.loc[[4,5],["Name","age"]]
```

##Conditional Selection

```python
df3
```
