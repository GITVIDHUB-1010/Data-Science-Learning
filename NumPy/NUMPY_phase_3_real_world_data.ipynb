#Real World Data with numpy

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 140000, 160000, 190000, 200000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])
print("===Zomato sales analysis===")
print("Shape of the data: ", sales_data.shape)
print("Sample data for 1st 3 restaurents:\n", sales_data[:3, :]) #[rows,columns]
```

###total sales per year

```python
print("Total sales of all the restaurent: \n",np.sum(sales_data, axis=0))
yearly_total= np.sum(sales_data[:, 1:],axis=0)
print("yearly total sales: \n",yearly_total )

#Minimum sales per restaurent
min_sales = np.min(sales_data[:,1:],axis=0)
print("minimum sales axis 0:\n",min_sales)

min_sales = np.min(sales_data[:,1:],axis=1)
print("minimum sales axis 1:\n",min_sales)

max_sales= np.max(sales_data[:,1:],axis=0)
print("maximum sales a:\n",max_sales)

avg_sales= np.mean(sales_data[:,1:],axis=1)
print('average sales of all the restaurents:\n',avg_sales)



```

```python
#Cumulative sales
cumsum= np.cumsum(sales_data[:,1:,],axis=1)
print(cumsum)

plt.figure(figsize=(6,5))
plt.plot(np.mean(cumsum,axis=0))
plt.title("Average cumulative sales across all the rest")
plt.xlabel("Years")
plt.ylabel('Sales')
plt.grid(True)
plt.show()
```
