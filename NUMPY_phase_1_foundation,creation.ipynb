# Numpy Basics - August 2026

```python
import numpy as np
```

###creating arrays from list

```python
arr_1d=np.array([[1,2,3,4,5]])
print("1D ARRAY: ",arr_1d)

arr_2d=np.array([[1,2,3],[4,5,6]])
print("2D ARRAY:", arr_2d)
```

###list vs numpy array

```python
py_list = [1,2,3]
print("list multiplication",py_list *3)

arr= np.array([1,2,3,4,5]) #element wise multiplication
print("array multiplication", arr**2)

import time
start = time.time()
py_list = [i*2 for i in range(10000)]
print("list operation", time.time()-start)

start = time.time()
arr= np.arange(10000)*2
print("arrays operation",time.time()-start)
```

###creating arrays from scratch

```python
zeros=np.zeros((3,4))
print("zeros array:\n",zeros)

ones=np.ones((3,4))
print("ones array:\n",ones)

full=np.full((3,3),5)
print("5 array full: \n",full)

random=np.random.random((2,3))
print("random: \n",random)

sequence = np.arange(0,11,1)
print('sequence range: \n',sequence)
```

##VECTOR, MATRIX AND TENSOR

```python
vector=np.array([1,2,3,4,5])
print("Vector:\n",vector)

matrix=np.array([[1,2,3],
                 [3,4,5],
                 [4,5,6]])
print("Matrix: \n",matrix)

tensor = np.array([[[1,2],[3,4]],
                   [[4,5],[5,6]]])
print("Tensor:\n",tensor)
```

###Array properties

```python
arr=np.array([[[1,2],[3,4]],
              [[4,5],[6,2]]])
print("shape: ",arr.shape)
print("dimensions: ",arr.ndim)
print("size: ",arr.size)
print("data type: ",arr.dtype)
```

###Array Reshaping

```python
arr = np.arange(20)
print("org array: ",arr)

res= arr.reshape((4,5))
print("reshaped array:\n ",res)

flat=res.flatten()
print("flattened array: \n",flat)

transp= res.T ##converting rows to column and columns to rows
print('transpose: \n',transp)
```
