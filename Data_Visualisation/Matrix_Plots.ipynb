##HeatMap
###Pivot table Heatmap


```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips=sns.load_dataset("tips")
tips
```

```python
tipscorr = tips[["total_bill","tip","size"]]
```

```python
tipscorr #correlation
```

```python
corr=tipscorr.corr()
corr
```

```python
sns.heatmap(corr,annot=True)
```

##Cluster Map

```python
sns.clustermap(corr,annot=True)
```

##pivot table heatmap

```python
fl=sns.load_dataset("flights")
fl
```

```python
pvt = fl.pivot_table(values="passengers",index="month",columns="year")
pvt
```

```python
sns.heatmap(pvt,cmap="Spectral")
```
