```python
import seaborn as sns
import matplotlib.pyplot as plt
```

###We use Distribution plots to visualize the distribution of **quantitative data**

```python
tips = sns.load_dataset('tips')
```

```python
tips
df = tips
```

```python
df["size"].unique() # there are only 6 values in the size column, so we consider this as categorical values
```

##Distplot/Histplot

```python
plt.subplot(2,2,1)
sns.histplot(df["total_bill"],bins=20,kde=True)
plt.subplot(2,2,2)
sns.histplot(df["tip"],bins=20,kde=True)
plt.subplot(2,2,3)
sns.histplot(df["size"],bins=20,kde=False)
```

##Joint Plot

```python
sns.jointplot(x="total_bill",y="tip",data =df,kind= "resid",color="magenta")
```

##Pair plot

```python
sns.pairplot(df,hue="day",palette="coolwarm")
```

##Rug Plot

```python
sns.rugplot(df,height =0.2,x="tip",y="total_bill")
```
