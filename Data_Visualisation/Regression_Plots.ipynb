```python
import seaborn as sns
import matplotlib.pyplot as plt
```

#Lmplot- linear model plot, used for linear regression

```python
tips =sns.load_dataset("tips")
tips
```

```python
sns.lmplot(data=tips,x="total_bill",y="tip",hue="sex",palette="magma",) #used in linear regression algorithm
```

```python
sns.lmplot(data=tips,x="total_bill",y="tip",hue="sex",palette="deep",markers=["o","v"],scatter_kws={'s':50}) #used in linear regression algorithm
```
