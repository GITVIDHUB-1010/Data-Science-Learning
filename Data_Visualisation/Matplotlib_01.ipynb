```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,5,11) #number of values from start to end, and how many values inside that
y = x**2
y
```

```python
plt.plot(x,y)
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("X and Y")
```

##Subplots in matplotlib

```python
plt.subplot(2,2,1)
plt.plot(x)
plt.subplot(2,2,2)
plt.plot(y)
plt.subplot(2,2,3)
plt.plot(x*y)
plt.subplot(2,2,4)
plt.plot(x**y)
```

##Object Oriented Way

```python
fig = plt.figure(figsize=(20,8),dpi=100)
axis1 = fig.add_axes([0.5,0.1,0.2,0.2])
axis1.plot(x,y)
plt.title("axis 1")
axis2 = fig.add_axes([0.5,0.4,0.2,0.2])
axis2.plot(x,y)
plt.title("axis 2")
# we can arrange the plots in our way, like custom places of our choice
```

##Type of plots

```python
plt.scatter(x,y)
```

```python
plt.hist(y)
```

```python
plt.boxplot(y)
```

#saving plots

```python
fig = plt.figure()
axis=fig.add_axes([0.1,0.1,0.8,0.8])
axis.plot(x,y)
plt.savefig("random plot.png")
```
