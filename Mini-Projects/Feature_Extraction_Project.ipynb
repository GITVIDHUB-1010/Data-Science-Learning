```python
def ext_epi(txt):
  check = False ## check works as a switch
  data = ""
  for i in txt:
    if i ==")":
      check = False
      return data
    if check == True:
      data += i
    if i == "(":
      check = True
```

```python
dfa["episodes"]= dfa["title"].apply(ext_epi)
dfa
dfa
```

```python
dfa["episodes"]=dfa['episodes'].astype(int)
```
